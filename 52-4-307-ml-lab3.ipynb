# Create a dictionary with 5 employee information : Ex - emp_ID, Post, SalaryPerHour, Bonus, WorkingHour

1. Add a totalSalary column with calculating total salary based on working hour.
2. Add a average row with calculating average salary of the employees.
3. Add a totalPayableAmount column with calculating sum of totalSalary and Bonus columns

```python
a = [20 , 40 , 30 , 20 , 10 ]
```

```python
import pandas as pd


a = {
    'emp_ID': ['1', '2', '3', '4', '5'],'Post': ['Manager','Developer','Analyst','Designer','QA'],'SalaryPerHour': [50, 40, 35, 35, 30],'Bonus': [500, 300, 200, 200, 150],
    'WorkingHour': [160, 150, 160, 140, 150]
}


df = pd.DataFrame(a)


df['totalSalary'] = df['SalaryPerHour'] * df['WorkingHour']

df['totalPayableAmount'] = df['totalSalary'] + df['Bonus']

avg = df[['SalaryPerHour', 'Bonus', 'WorkingHour', 'totalSalary', 'totalPayableAmount']].mean()


avg['emp_ID'] = 'Average'
avg['Post'] = '-'


df.loc['Avg_Row'] = avg


print(df)
```
